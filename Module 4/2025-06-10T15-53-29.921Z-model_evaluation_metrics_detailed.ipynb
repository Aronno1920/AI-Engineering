#  Introduction to Model Evaluation Metrics in Regression
In this notebook, we’ll learn how to evaluate the accuracy of machine learning models that predict **numerical values** (also known as **regression models**).

We'll cover:
- What is **Mean Absolute Error (MAE)**?
- What is **Root Mean Squared Error (RMSE)**?
- What is **R-squared (R²)**?
- How to interpret and choose between these metrics

##  Step 1: Import Required Libraries

```python
# NumPy is used for working with arrays (lists of numbers)
import numpy as np

# Scikit-learn provides functions to calculate model evaluation metrics
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

##  Step 2: Define the Actual and Predicted Values

```python
# These are the correct values that a model should have predicted
actual_values = np.array([20, 30, 40, 60, 70, 80, 90])

# Model 1 makes very poor predictions (just random numbers to show what a bad model looks like)
predicted_model_1 = np.array([324234, 345, 435, 43534, 345, 435345, 45645])

# Model 2 makes very good predictions (close to actual values)
predicted_model_2 = np.array([21, 30, 41, 61, 70, 81, 90])
```

##  Step 3: Calculate Mean Absolute Error (MAE)

```python
# Calculate MAE for each model
mae_model_1 = mean_absolute_error(actual_values, predicted_model_1)
mae_model_2 = mean_absolute_error(actual_values, predicted_model_2)

# Print the results
print("Mean Absolute Error (Model 1):", mae_model_1)
print("Mean Absolute Error (Model 2):", mae_model_2)
```

 **What is MAE?**  
MAE is the **average** of the **absolute differences** between the actual and predicted values.  
- It tells us how far off our predictions are, on average.  
- **Lower MAE = better model**

##  Step 4: Calculate Root Mean Squared Error (RMSE)

```python
# Calculate RMSE for each model
rmse_model_1 = mean_squared_error(actual_values, predicted_model_1, squared=False)
rmse_model_2 = mean_squared_error(actual_values, predicted_model_2, squared=False)

# Print the results
print("Root Mean Squared Error (Model 1):", rmse_model_1)
print("Root Mean Squared Error (Model 2):", rmse_model_2)
```

 **What is RMSE?**  
RMSE is the **square root** of the average of squared differences between predicted and actual values.  
- It **penalizes large errors more** than MAE.  
- Good when we care more about big mistakes.  
- **Lower RMSE = better model**

##  Step 5: Calculate R-squared (R² Score)

```python
# Calculate R² Score for each model
r2_model_1 = r2_score(actual_values, predicted_model_1)
r2_model_2 = r2_score(actual_values, predicted_model_2)

# Print the results
print("R-squared (Model 1):", r2_model_1)
print("R-squared (Model 2):", r2_model_2)
```

 **What is R²? (R-squared)**  
R² tells us **how well the model explains the actual data**.  
- **1.0 = perfect predictions**  
- **0.0 = as good as guessing the average**  
- **Negative = worse than guessing**

##  Step 6: Summary – Choosing the Right Metric

| Metric | Use When... | Notes |
|--------|-------------|-------|
| MAE    | You want to treat all errors equally | Easy to understand |
| RMSE   | You want to punish larger errors more | More sensitive to outliers |
| R²     | You want to measure how well the model fits overall | Works only for regression models |
